##### Type casting

`Type Casting` : 
1. `Upcasting` : converting a smaller/lower data type into a larger/higher data type automatically.
    - ptyhon does this to avoid data loss.
2. `Downcasting` : convertn a higher data type into a lower data type mannually.
    - It may cause data loss.

```python
# 1. upcasting
# data type from lower to gigher: int -> float -> complex

a =10 #int
b = 10.5 #float
print(a+b) # float(higer data type)
type(a+b)
```

```python
# Downcasting
# data type from higher to lower: complex -> float -> int
a = 10.5
print(type(a))
b = int(a)
print(b)
print(type(b))


```

`List`

```python
# List: Ordered(maintain insertion order), mutable collection, dublicate allowed, support indexing and slicing and can store diff data types.
num1 = [10,20,30,40,10]
list1 = [10,"Ai",100,12.5, True]
print(num1)
print(list1)
print(type(num1))
print(type(list1))
```

```python
# update list1
num1[1] = 40
print(num1)
```

```python
# add a new elemet at the end of list
num1.append(50)
print(num1)
```

```python
# delete element from list
num1.remove(10)
print(num1)
```

`Tupple`

```python
# Tuple: Orderd but immutable collection.
num = (10,20,30,40,50)
print(num)
type(num)
```

```python
# # update not possible because tuple is immutable
num[1] = 100
print(num)
# update not possible because tuple is immutable
```

```python
# delete - we can not delete a single element from tuple but we can delete entire tuple
a = (10,20,30,40,50)
del a[[1]]
print(a)
```

```python
# delete - we can delete entire tuple
del a 
print(a)
```

`Set`

```python
# Set: Unorderd collection, No dublicate value.
num = {10,20,30,40,50}
print(num)
type(num)
```

```python
# update
num.add(60)
print(num)
```

```python
# delete
num.remove(50)
print(num)
```

`Dictionary`

```python
# Dictionary: store data in key : value pairs, Unordered collection, mutable, no dublicate key allowed, fast access of data using key.
emp = {
    "name":"Ahishek",
    "exp":5,
    "city":"Dwaraka, New Delhi"
}
print(emp)
type(emp)
```

```python
# update
emp["job"] = 'Gen Ai Engineer'
print(emp)
```

```python
emp['lorem'] = 'lorem'
print(emp)
```

```python
# delete
del emp['lorem']
print(emp)
```
