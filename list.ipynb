##### List
- List is an ordered,mutable(changable) collection in python that allows duplicate values and can store dufferent types of data. It support CRUD operation.

- positive index: start from 0.
- negative index: start from -1

```python
# CRUD(create, read, update, delete)
#  Create
rough_list = ["Mango", "Banana",100,10.5,True]


# Read
print(rough_list)
print(rough_list[-2])
print(type(rough_list))
# For access of all value use :
print(": ",rough_list[:])

# Update : use close bracket for indexing using nums+ve , -ve or 0.
rough_list[4] = "New Delhi"
print(rough_list)

# Delete
rough_list.remove(10.5)
rough_list.remove("Banana")
print(rough_list)

# Use pop() to remove last element
rough_list.pop()
print(rough_list)

```

- List slicing : retriving specific data/ a part from list.
seq[start : stop : step]

start: start index
stop: ending index
step: num of position to skip(optional)

|Syntax|Meaning|
|-------|-------|
|list[-1:]|Last element to end ->[true]|
|list[:1]|From beginning to last element(last excluded)|
|list[-2:]|Last 2 elements|
|list[:-2]|All expect last 2 elements|

```python
rough_list2 = ["Mango", "Banana",100,10.5,True]

# slicing - copy/return complete list  :: or :
print(rough_list2[::])


# basic slicing - print starting index to ending index list :
print(rough_list2[1:4])

# slice from beginning :number
list2 = ["Mango", "Banana",100,10.5,True]
print("Original List",list2[::])
print("2: list starting with ind 2 then next words", list2[2:])

# print in reverse maanner
print(list2[::-1])
```

```python
# slice till end number:
list2 = ["Mango", "Banana",100,10.5,True]
print(list2[1:])
```

```python
list2 = ["Mango", "Banana",100,10.5,True]
list3 = ["Mango", "Banana",100,10.5,True]
list4 = ["Mango", "Banana",100,10.5,True]

# Negative index from end -num: or :num-
print(list2[-1:])

# -ve index from starting
print(list3[:-1])
```

- Lists are mutable, meaning we can replace the value at a specific index, add new elements, or remove existing elements without creating a new list.

```python
lists3 = ["Mango","Banana", "Apple","Graps"]
print(lists3)
print(lists3[2])
#replace value of index 2
lists3[2] = "Lichi"
print(lists3)
# Looping statement
for items in lists3:
    print(items)
```

- List Methods
    - append() - add an element to the end of the list
    - extend() -  add multipple elements to the end of the list
    - insert() - add an element at a specific Index
    - remove() - remove an element from the list by value
    - pop() -  remove an elemt from the list at specific Index
    - clear() - remove all elements from list
    - index() - return the element of first matching value 
    - count() - count how many times a value appears in the list 
    - sort() - sort the list in ascending order
    - reverse() - reverse the order of the list 
    - copy() - creates a shallow copy of tje list 


```python
# append()
lst4 = ["Mango", "Banana", "Apple", "Lichi"]
lst4.append("Jamun")
print(lst4)
```

```python
# extend() : syntax
lst4 = ["Mango", "Banana", "Apple", "Lichi"]
print(lst4)
lst4.extend(["Strawberry", "Blueberry"])
print(lst4)
```

```python

# insert()
lst4 = ["Mango", "Banana", "Apple", "Lichi"]
print(lst4)
lst4.insert(2, "Papaya")
print(lst4)
```

```python
# remove()
lst4 = ["Mango", "Banana", "Apple", "Lichi"]
print(lst4)
lst4.remove("Banana")
print(lst4)
```

```python
# pop()
lst4 = ["Mango", "Banana", "Apple", "Lichi"]
print(lst4)
lst4.pop(3)
print((lst4))
```

```python
# clear()
lst4 = ["Mango", "Banana", "Apple", "Lichi"]
print(lst4)
lst4.clear()
print(lst4)
```

```python
# index()
lst4 = ["Mango", "Banana", "Apple", "Lichi"]
print(lst4)
print(lst4.index("Mango"))
```

```python
# count()
lst4 = ["Mango", "Banana", "Apple", "Lichi","Mango"]
print(lst4)
print(lst4.count("Mango"))
```

```python
# sort()
lst4 = [10,5,2,66,100,1000]
lst4.sort()
print(lst4)

# sort()
lst4 = ["Mango", "Banana", "Apple", "Lichi"]
print(lst4.sort())
print(lst4)
```

```python
# reverse()
lst4 = ["Mango", "Banana", "Apple", "Lichi"]
print(lst4)
lst4.reverse()
print(lst4)
```

```python
# copy()
# sort()
lst4 = ["Mango", "Banana", "Apple", "Lichi"]
print(lst4)
lst4.copy()
print(lst4)
```

- enumarete()

used to get bothh index and value while iterating over a sequence.

syntax: enumarate(iterable, start=0)

```python
# enumerate()
lst4 = ["Mango", "Banana", "Apple", "Lichi"]
for i, item in enumerate(lst4):
    print(i, item)
```

```python
# enumerate(): start from 1
lst4 = ["Mango", "Banana", "Apple", "Lichi"]
for i, item in enumerate(lst4, start=1):
    print(i, item)
```

- comprehension() 

a concise way to create a new list by applying an expression to each item in an existing iterable and optionaly filtering item based on consition.

Syntax:

[expression for item in iterable]

```python
# comprehension
square = [x*x for x in range(1,6 )]
print(square)
```

```python
# compehension print even numbers
even_numbers = [x for x in range(1, 16) if x%2 == 0]
print(even_numbers)
```

```python
# comprehension with if condition
even_square = [x*x for x in range(1,11) if x%2==0]
print(even_square)
```

- zip()

combines multiple iterables.

```python
# zip() function
emp = ["Accenture", "TCS", "HCL"]
no_of_devv = [300, 400, 900]
department = ["AI Engineer", "GenAI Developer", "Data Engineer"]
for a,b,c in zip(emp, no_of_devv, department):
    print(a, b, c)
```
